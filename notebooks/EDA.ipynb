```python
import pandas as pd

df = pd.read_csv("../data/pakwheels_used_cars.csv")
```

```python
print(df["price"].quantile([0.01, 0.05, 0.5, 0.95, 0.99]))
print(df["mileage"].quantile([0.01, 0.05, 0.5, 0.95, 0.99]))
print(df["mileage"].value_counts().head(10))
print(df["engine_cc"].value_counts().head(15))
print(df[["make", "model", "ad_city", "registered", "transmission"]].nunique())
print(df["fuel_type"].value_counts())
print(df["make"].value_counts().head(10))
```

```python
df.columns
```

```python
df = df.drop(columns=["ad_ref", "year"])
```

```python
df.columns
```

```python
df = df.dropna(subset=["price", "engine_cc"])
```

```python
df.isnull().sum()
```

```python
for col in ["assembly", "body", "color", "fuel_type"]:
    df[col] = df[col].fillna("Unknown")
```

```python
df[["assembly", "body", "color", "fuel_type"]].isnull().sum()
```

```python
n_start = len(df)

print(n_start)
```

```python
print("Rows before:", len(df))

df = df[df["mileage"] != 123456]

print("Rows after:", len(df))
```

```python
print("Rows before:", len(df))

df = df[df["mileage"].between(100, 400000)]

print("Rows after:", len(df))
```

```python
print("Rows before:", len(df))

df = df[df["engine_cc"].between(600, 6000)]

print("Rows after:", len(df))
```

```python
lo, hi = df["price"].quantile([0.01, 0.99])

print("Lower limit:", lo)
print("Upper limit:", hi)
```

```python
print("Rows before:", len(df))

df = df[df["price"].between(lo, hi)]

print("Rows after:", len(df))
```

```python
print(f"Started with {n_start} rows")
print(f"Ended with {len(df)} rows")
print(f"Kept {len(df) / n_start:.1%} of the data")
```

```python
df.to_csv("../data/cleaned_cars.csv", index=False)
```

# Exploratory Data Analysis

## Price Distribution

```python
import pandas as pd 
df = pd.read_csv("../data/cleaned_cars.csv")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,5))

sns.histplot(df["price"], bins=50, kde=True)

plt.title("Distribution of Car Prices")
plt.xlabel("Price (PKR)")
plt.ylabel("Number of Cars")

plt.show()
```

```python
df["price"].max()
```

```python
plt.figure(figsize=(10,6))

top_makes = df["make"].value_counts().head(10)

sns.barplot(x=top_makes.values, y=top_makes.index)

plt.title("Top 10 Car Makes")
plt.xlabel("Number of Listings")
plt.ylabel("Car Make")

plt.show()
```

### Observation

Toyota, Suzuki, and Honda dominate the dataset. This indicates that the Pakistani used-car market is heavily concentrated around a few manufacturers.

```python
plt.figure(figsize=(10,6))

top_cities = df["ad_city"].value_counts().head(10)

sns.barplot(x=top_cities.values, y=top_cities.index)

plt.title("Top 10 Cities by Listings")
plt.xlabel("Number of Listings")
plt.ylabel("City")

plt.show()
```

### Observation

Most listings come from major cities, reflecting larger used-car markets and higher online marketplace activity.

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df, x="fuel_type")

plt.title("Fuel Type Distribution")
plt.xlabel("Fuel Type")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.countplot(data=df, x="transmission")

plt.title("Transmission Distribution")
plt.xlabel("Transmission")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.countplot(data=df, x="assembly")

plt.title("Assembly Type")
plt.xlabel("Assembly")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(10,6))

top_body = df["body"].value_counts().head(10)

sns.barplot(x=top_body.values, y=top_body.index)

plt.title("Top 10 Body Types")
plt.xlabel("Number of Listings")
plt.ylabel("Body Type")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(df["engine_cc"], bins=40, kde=True)

plt.title("Engine Size Distribution")
plt.xlabel("Engine Size (cc)")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(df["mileage"], bins=50, kde=True)

plt.title("Mileage Distribution")
plt.xlabel("Mileage (km)")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df.sample(5000, random_state=42),
    x="mileage",
    y="price",
    alpha=0.5
)

plt.title("Price vs Mileage")
plt.xlabel("Mileage (km)")
plt.ylabel("Price (PKR)")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df.sample(5000, random_state=42),
    x="engine_cc",
    y="price",
    alpha=0.5
)

plt.title("Price vs Engine Size")
plt.xlabel("Engine Size (cc)")
plt.ylabel("Price (PKR)")

plt.show()
```

```python
plt.figure(figsize=(6,5))

avg_price = df.groupby("transmission")["price"].mean().sort_values()

sns.barplot(
    x=avg_price.index,
    y=avg_price.values
)

plt.title("Average Price by Transmission")
plt.xlabel("Transmission")
plt.ylabel("Average Price (PKR)")

plt.show()
```

```python
plt.figure(figsize=(8,6))

numeric_df = df.select_dtypes(include=["int64", "float64"])

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(x=df["price"])

plt.title("Boxplot of Car Prices")

plt.show()
```
